### Scrub, listen, select

The clip preview is a remote control for kernel playback (same transport as `daw.play`):

- **click** the waveform to move the cursor (while playing, playback jumps there)
- **drag** to select a range, **shift-click** to select from the cursor
- **▶ / Enter** plays the selection (or from the cursor), **■ / Space** stops, **⟳** toggles loop, **Esc** clears
- the selection shows as pasteable slice syntax, e.g. `[0.512:1.034]`; **copy** puts it on the clipboard

Keep a handle with `take.view()` to read the selection back in code.

```python
v = take.view()
v
```

```python
v.range, daw.selection()
```

```python
hook = v.region()  # same node as typing take[a:b]
hook
```

```python
take.peak_db(), take.rms_db()
```
