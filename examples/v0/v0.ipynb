# baddaw v0: load, see, hear, export

Put your recording next to this notebook (or run `python v0.py` once to generate `demo.wav`).

```python
import baddaw as daw
```

```python
take = daw.load("demo.wav")
take
```

```python
take.peak_db(), take.rms_db()
```

```python
loop = take[0.5:1.0]
daw.play(loop, loop=True)
```

```python
daw.status()
```

```python
daw.stop()
```

```python
quiet = take.gain(-6)
quiet
```

```python
quiet.export("quiet.wav")
```
